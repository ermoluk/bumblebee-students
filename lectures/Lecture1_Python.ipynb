## 1. `print()` - make the computer talk
`print()` shows text on the screen. Text always goes inside quotes `"like this"`.

```python
print("Hello!")
print("I am learning Python")
```

## 2. Variables - boxes that remember
A **variable** is a box with a name. You put a value inside with `=`, then use the name later.
`=` means *"store into"*, not *"equals"* like in math.

```python
# put 1.0 into a box called height
height = 1.0
name = "Bumblebee"

print(height)
print(name)
```

## 3. Types - kinds of values
Every value has a **type**. `type(...)` tells you which one.
- Number (integer): `5, 42, -3`
- Number (decimal): `1.0, 0.4, 3.14`
- Text (string): `"hello"` — always in quotes
- True / False (boolean) — yes-or-no answers

```python
print(5,       type(5))
print(1.0,     type(1.0))
print("hello", type("hello"))
print(True,    type(True))
```

## 4. Comments — notes for humans
Anything after `#` is a comment. Python ignores it — comments explain **why**.

```python
# This line is a note. Python ignores it.
speed = 0.4      # comments can sit after code too
print(speed)     # only the code runs
```

## 5. Doing math
Python is a calculator too.

```python
a = 10
b = 3

print(a + b)   # add
print(a - b)   # subtract
print(a * b)   # multiply
print(a / b)   # divide
```

## 6. `if` / `else` — making decisions
If a question is **True** → do one thing, `else` → do another.
Notice the `:` and the **indent** (spaces) — they tell Python what belongs to the `if`.

```python
battery = 30

if battery > 20:
    print("OK to fly")
else:
    print("Charge me first!")
```

## 7. `for` — repeat things
A loop repeats the same steps. `range(3)` gives us `0, 1, 2`. No copy-paste needed!

```python
for i in range(3):
    print("Beep", i)
```

## 8. Functions — teach a new word
A function is a mini-program with a name. `def` creates it; you **call** it by its name.
Write once, use many times. (Later, `drone.takeoff()` will just be a function someone wrote for us.)

```python
def say_hi(name):
    print("Hi", name)

say_hi("Anna")
say_hi("Max")
```

## 9. Putting it together
Variables + loop + function, all at once.

```python
# A tiny program using everything we learned
def countdown(start):
    for i in range(start, 0, -1):
        print(i)
    print("Lift off!")

countdown(3)
```
